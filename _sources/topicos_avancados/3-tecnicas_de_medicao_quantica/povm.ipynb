Uma POVM é um conjunto de operadores positivos ${E_i}$ tal que:

$$
\sum_i{E_i} = I	​
$$

E a probabilidade de obter o resultado i, dado um estado ρ, é:

$$
p(i)=Tr(E_i ρ)
$$

```python
# MÉTODO DIRETO - cacula com Tr(E_i * rho) para cada E_i


def direct_povm(rho, povm_elements, return_probs=False):
    """
    Simula uma medição POVM pelo método direto

    Args:
        rho (np.ndarray): estado quântico (matriz densidade)
        povm_elements (list): lista de elementos POVM (matrizes positivas)
        return_probs (bool): se True, retorna também as probabilidades associadas a cada resultado

    Returns:
        result (int): índice do resultado da medição
        probs (np.ndarray, opcional): probabilidades associadas a cada resultado (se return
    """

    probs = np.array([np.trace(E @ rho) for E in povm_elements])

    # estabilidade numérica
    probs = np.maximum(probs, 0)
    probs /= probs.sum()

    result = np.random.choice(len(povm_elements), p=probs)

    if return_probs:
        return result, probs

    return result
```

```python
from IPython.display import display, HTML
import ipywidgets as widgets
import pandas as pd

# =========================
# Perguntas do questionário
# =========================

perguntas = [
    "A plataforma foi intuitiva.",
    "Os exemplos facilitaram a compreensão.",
    "Os exercícios ajudaram no aprendizado.",
    "O conteúdo estava bem organizado.",
    "Eu utilizaria a plataforma novamente.",
]

# Escala Likert
escala = [
    ("Discordo totalmente", 1),
    ("Discordo", 2),
    ("Neutro", 3),
    ("Concordo", 4),
    ("Concordo totalmente", 5),
]

# =========================
# Interface
# =========================

display(HTML("<h2>Questionário de Avaliação</h2>"))
display(HTML("<p>Selecione uma opção para cada pergunta.</p>"))

widgets_perguntas = []

for i, pergunta in enumerate(perguntas):

    radio = widgets.RadioButtons(
        options=escala, description="", layout=widgets.Layout(width="500px")
    )

    widgets_perguntas.append(radio)

    display(HTML(f"<b>{i+1}. {pergunta}</b>"))
    display(radio)

# =========================
# Botão de envio
# =========================

botao = widgets.Button(description="Enviar", button_style="success")

output = widgets.Output()

display(botao)
display(output)

# =========================
# Função de envio
# =========================


def enviar_questionario(b):

    respostas = []

    for pergunta, widget in zip(perguntas, widgets_perguntas):
        respostas.append({"Pergunta": pergunta, "Resposta": widget.value})

    df = pd.DataFrame(respostas)

    media = df["Resposta"].mean()

    with output:
        output.clear_output()

        print("Questionário enviado com sucesso!\n")

        display(df)

        print(f"Média geral: {media:.2f}")


botao.on_click(enviar_questionario)
```
